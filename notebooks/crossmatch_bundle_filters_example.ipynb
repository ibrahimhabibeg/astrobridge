```python
from astrobridge.paper_pairing.core import CrossmatchBundle
from astrobridge.paper_pairing.filters import (
    MaxObjectsPerPaperFilter,
    MaxPapersPerObjectFilter,
    ObjectNameInTitleOrAbstractFilter,
    TitleKeywordFilter,
    ObjectNameOrAliasInTitleOrAbstractFilter
)
from astrobridge.paper_pairing.augmenters import AliasAugmenter
```

```python
bundle = CrossmatchBundle.load("../data/paper_crossmatch_bundles/mmu_desi_edr_sv3")
print(bundle.summary())
```

```python
filter = TitleKeywordFilter()
filtered_bundle = filter.apply(bundle, inplace=False)
print(filtered_bundle.summary())
```

```python
filter = ObjectNameInTitleOrAbstractFilter()
filtered_bundle = filter.apply(bundle, inplace=False)
print(filtered_bundle.summary())
```

```python
filter = MaxObjectsPerPaperFilter(max_objects=1)
filtered_bundle = filter.apply(bundle, inplace=False)
print(filtered_bundle.summary())
```

```python
filter = MaxPapersPerObjectFilter(max_papers=1)
filtered_bundle = filter.apply(bundle, inplace=False)
print(filtered_bundle.summary())
```

```python
# augmenter = AliasAugmenter()
augmented_bundle = CrossmatchBundle.load("../data/paper_crossmatch_bundles/mmu_desi_edr_sv3_augmented")
print(augmented_bundle.summary())
filter = ObjectNameOrAliasInTitleOrAbstractFilter()
filtered_bundle = filter.apply(augmented_bundle, inplace=False)
print(filtered_bundle.summary())
```
